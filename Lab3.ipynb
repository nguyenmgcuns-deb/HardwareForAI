```python
from sklearn import datasets
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.layers import Input, Dense 
from tensorflow.keras.models import Model
from tensorflow.keras.models import load_model
```

```python
Step 3: Dataset analysis
```

```python
diabetes = datasets.load_diabetes()
```

```python
diabetes['feature_names']
```

```python
features = diabetes.data[:,2:3]
```

```python
len(features)
```

```python
print(features.shape)         
print(np.squeeze(features).shape) 
```

```python
Step 4: Plot dataset
```

```python
targets = diabetes.target
```

```python
plt.scatter(features,targets)
plt.xlabel("BMI")
plt.ylabel("Disease progression")
```

```python
Step 5: Neuron preparation
```

```python
inputs = Input(shape=(1,))
output = Dense(1, activation='linear')(inputs)
```

```python
linear_model = Model(inputs, output)
```

```python
linear_model.summary()
```

```python
linear_model.compile(optimizer='sgd', loss='mse')
```

```python
# model training with features and targets in dataset for a limited number of epochs
EPOCH = 1000
r = linear_model.fit(features, targets, epochs=EPOCH)
```

```python
# plot the statistics of training
x = np.linspace(1,EPOCH,EPOCH)
plt.plot(x, r.history['loss'],'o', label='loss')
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
```

```python
# Plot the transfer function of the trained neuron
pred = linear_model.predict(features)
```

```python
# plot the function of neuron by definition
weight = linear_model.get_weights()[0].item()
bias = linear_model.get_weights()[1].item()

plt.scatter(features,targets, label = 'raw data')
plt.plot(features, weight * features + bias, 'r', label = 'transfer function y = wx+b')
plt.xlabel("BMI")
plt.ylabel("Disease progression")
plt.legend()
```

```python
Step 6: Training preparation
```

```python
# introduce the new feature, the experiments with **BMI** and **blood pressure**
features_2d = diabetes.data[:, 2:4]
```

```python
# observing the correlation between BP and progression

x = features_2d[:, 1]
y = targets
plt.xlabel("Blood Pressure")
plt.ylabel("Disease progression")

plt.scatter(x,y)
```

```python
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Lấy cột đầu tiên (BMI) cho trục x và cột thứ hai (Blood pressure) cho trục y
x = features_2d[:, 0]
y = features_2d[:, 1]
z = targets

ax.set_xlabel('BMI')
ax.set_ylabel('Blood pressure')
ax.set_zlabel('Disease progression')

ax.scatter(x, y, z, marker='o')
```

```python
# implement the single neuron with tensorflow
# Đầu vào bây giờ có kích thước là 2 (gồm BMI và BP)
inputs = Input(shape=(2,))
output = Dense(1, activation='linear')(inputs)
```

```python
# construct the model
# The funcaction "Model" is imported from tensorflow
linear_model_bmi_bp = Model(inputs, output)
```

```python


# check details of the model
linear_model_bmi_bp.summary()
```

```python
# compile the model with tensorflow with predefined optimizer and loss function
linear_model_bmi_bp.compile(optimizer='sgd', loss='mse')
```

```python
# model training with features and targets in dataset
EPOCH = 1000
# Truyền cả đầu vào 2D và đích vào đây
r_2d = linear_model_bmi_bp.fit(features_2d, targets, epochs=EPOCH)
```

```python
# plot the statistics of training and compare the results between two models
x = np.linspace(1,EPOCH,EPOCH)
# Điền lịch sử loss của mô hình 1 biến (r) và mô hình 2 biến (r_2d)
plt.plot(x, r.history['loss'],'o', label='loss_bmi')
plt.plot(x, r_2d.history['loss'],'o', label='loss_bmi_bp')

plt.xlabel("Epoch")
plt.ylabel("Loss")

plt.legend()
```

```python
# plot the statistics of training and compare the results
x = np.linspace(1,EPOCH,EPOCH)
# Cắt lát dữ liệu loss từ epoch 100 đến 1000
plt.plot(x[100:1000], r.history['loss'][100:1000],'o', label='loss_bmi')
plt.plot(x[100:1000], r_2d.history['loss'][100:1000],'o', label='loss_bmi_bp')

plt.xlabel("Epoch")
plt.ylabel("Loss")

plt.legend()
```

```python
# observing the transfer function
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

weight_1 = linear_model_bmi_bp.get_weights()[0][0].item()
weight_2 = linear_model_bmi_bp.get_weights()[0][1].item()
bias = linear_model_bmi_bp.get_weights()[1].item()


# Tách cột dữ liệu x, y từ features_2d
x = features_2d[:, 0]
y = features_2d[:, 1]
z = targets
xx = np.linspace(-0.1,0.15,1000)
yy = np.linspace(-0.1,0.1,1000)

ax.set_xlabel('BMI')
ax.set_ylabel('Blood pressure')
ax.set_zlabel('Disease progression')


ax.scatter(x, y, z, marker='o', label = 'raw data')
# Điền công thức w1*x + w2*y + b
ax.plot(x, y, weight_1 * x + weight_2 * y + bias, 'o', color= 'orange', label = 'transfer function')
# Điền công thức w1*xx + w2*yy + b cho lưới quét (sweep)
ax.plot(xx, yy, weight_1 * xx + weight_2 * yy + bias, color= 'r', label = 'transfer function sweep')

plt.legend()
```
